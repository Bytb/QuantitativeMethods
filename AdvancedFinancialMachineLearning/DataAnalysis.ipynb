```python
#imports
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tickers = ['MSFT', 'LLY', 'WMT', 'XOM', 'BA']
test = [yf.Ticker(ticker).history(period='1y')['Close'] for ticker in tickers]
#making pandas df
df = pd.concat([yf.Ticker(ticker).history(period='1y')['Close'] for ticker in tickers],axis=1)
df.columns = tickers

df.plot(kind='line', ax=None)
plt.show()
```

```python
covMatrix = df.cov()
print(covMatrix)
```
