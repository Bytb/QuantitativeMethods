```python
#imports
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sci_functions import Covariance_PCA
```

```python
tickers = ['MSFT', 'LLY', 'WMT', 'XOM', 'BA','GM','MCD','META','GOOG','NIO']
test = [yf.Ticker(ticker).history(period='1y')['Close'] for ticker in tickers]
#making pandas df
df = pd.concat([yf.Ticker(ticker).history(period='1y')['Close'] for ticker in tickers],axis=1)
df.columns = tickers

df.plot(kind='line', ax=None)
plt.show()
```

```python
covMatrix = df.cov()
eigenvalues, eigenvectors, variance_explained = Covariance_PCA(
    covMatrix,
    n_components=2  # Change to 2 for a 2D plot
)

# Example: coefficients for every asset in PC1
print(eigenvectors[:, 0])

# Example: percentage of total asset variance explained by PC1
print(variance_explained[0])
```
